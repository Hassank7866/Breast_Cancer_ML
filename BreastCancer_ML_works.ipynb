```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report
)

import joblib
```

```python
dataset = load_breast_cancer(as_frame=True)

X = dataset.data
y = dataset.target

print("Features shape:", X.shape)
print("Labels shape:", y.shape)

print("\nFirst five rows:")
print(X.head())

print("\nTarget classes:")
print(dataset.target_names)

print("\nClass counts:")
print(y.value_counts())
```

```python
print("Missing values:", X.isnull().sum().sum())
print("Duplicate rows:", X.duplicated().sum())

print("\nFeature statistics:")
print(X.describe().round(2)) #dekh rahe kitni missing values and duplicate rows hein hmare
#piare dataset me
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=2000, random_state=42)
)

model.fit(X_train, y_train)

print("Model trained successfully!")
```

```python
y_pred = model.predict(X_test)

print("Actual labels:   ", y_test.iloc[:10].to_list())
print("Predicted labels:", y_pred[:10].tolist())
```

```python
# Calculate metrics on the entire test set
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"Overall Accuracy: {accuracy:.4f} ({accuracy * 100:.2f}%)")
print(f"Precision:        {precision:.4f}")
print(f"Recall:           {recall:.4f}")
print(f"F1-Score:         {f1:.4f}")

# Let's plot the Confusion Matrix to see exactly where any mistakes occurred
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=dataset.target_names)
disp.plot(cmap=plt.cm.Blues)
plt.title("Confusion Matrix")
plt.show()
```

### What Measurements Are We Looking At?

This dataset (the **Wisconsin Breast Cancer Dataset**) contains measurements taken from a **fine needle aspirate (FNA)**—essentially, a tiny biopsy sample of a breast lump.

Under a microscope, we measure **10 key physical characteristics** of the cell nuclei in the sample:

1. **Radius**: The average distance from the center to the outer edge of the cell nucleus.
2. **Texture**: The standard deviation of gray-scale values (how rough or smooth the nucleus looks under imaging).
3. **Perimeter**: The distance all the way around the outside of the nucleus.
4. **Area**: The total size of the nucleus.
5. **Smoothness**: How much the radius length varies locally.
6. **Compactness**: Computed as $\frac{\text{perimeter}^2}{\text{area}} - 1.0$ (how tightly packed the nucleus shape is).
7. **Concavity**: The severity of indentations or "caves" in the boundary of the nucleus.
8. **Concave Points**: The number of these indentations.
9. **Symmetry**: How symmetrical the nucleus shape is.
10. **Fractal Dimension**: An index measuring how complex/rough the boundary line of the nucleus is.

### How We Get 30 Features
For each of these 10 physical characteristics, three values are calculated for every biopsy sample:
* **Mean**: The average value of that characteristic across all cells in the sample.
* **Standard Error (SE)**: The variation in that measurement across the cells.
* **Worst / Largest**: The average of the three largest values (the worst-case outliers).

$10 \text{ characteristics} \times 3 \text{ perspectives} = 30 \text{ total features}$.

### Why Logistic Regression Works So Well Here
Malignant (cancerous) cells tend to have much larger, darker, and more irregular/misshapen nuclei compared to healthy, benign cells. Because these physical features are so distinct, our Logistic Regression model can easily find a clear boundary to separate them with **98.25% accuracy**!

```python
print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))

print(
    "Precision (malignant):",
    round(precision_score(y_test, y_pred, pos_label=0), 4)
)

print(
    "Recall (malignant):",
    round(recall_score(y_test, y_pred, pos_label=0), 4)
)

print(
    "F1-score (malignant):",
    round(f1_score(y_test, y_pred, pos_label=0), 4)
)

print("\nFull classification report:")

print(classification_report(
    y_test,
    y_pred,
    target_names=dataset.target_names
))
```
